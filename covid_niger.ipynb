```python
# Clean text columns
df['grossesse'] = df['grossesse'].astype(str).str.lower().str.strip()
df['Si_oui_prcisez_le_trimestre'] = df['Si_oui_prcisez_le_trimestre'].astype(str).str.strip()

# Keep only pregnant women
dff = df[df['grossesse'] == 'oui']

# Remove missing trimester values
dff = dff[dff['Si_oui_prcisez_le_trimestre'].notna()]

# Plot distribution
sns.countplot(data=dff, x='Si_oui_prcisez_le_trimestre')
plt.xlabel('Trimestre de grossesse')
plt.ylabel('Nombre de cas')
plt.title('Distribution des cas par trimestre de grossesse')
plt.show()
```
